# Analysing the Data

```python
# 1- Plotting the Number of Active and Churn Customers
```

```python
ax=sns.countplot(x='Churn', data=df)
ax.bar_label(ax.containers[0])
plt.title("Bar Graph Showing Number of Active and Churned Customers\n", fontsize=20)
plt.show()
```

```python
bx=sns.countplot(x='gender', data=df, hue='Churn')
bx.bar_label(bx.containers[0])
bx.bar_label(bx.containers[1])
plt.show()
```

```python
 # 2-finding the churn pattern with respect to tenure
```

```python
sns.histplot(x='tenure', data=df, bins=30, hue='Churn')
plt.title("Plot of Count of Churn versus the tenure\n", fontsize=30)
plt.show()
```

```python
# 3- Plotting the Churn Pattern with respect to Contract 
```

```python
ax=sns.countplot(x='Contract', data=df, hue='Churn')
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.title("Bar Graph Showing Count of people based on contract\n", fontsize=20)
plt.show()
```

```python
churn_by_tenure=df[df['Churn']=='Yes'].groupby('Contract')['customerID'].count()
churn_by_tenure
```

```python
plt.pie(churn_by_tenure, labels=churn_by_tenure.index, autopct='%.0f%%')
plt.show()
```

```python
#conclusion- People with month to month contract tends to churn out more likely
```

```python
df.columns.values
```

```python
# 4- Finding the churn pattern with respect to Payment Method
```

```python
plt.figure(figsize=(10,5))
ax=sns.countplot(x='PaymentMethod', data=df, hue='Churn')
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.title("Bar Graph Showing Count of people Active and Churn based on Payment Method\n", fontsize=20)
plt.xticks(rotation=30)
plt.show()
```

```python
#conclusion - the customers are more likely to churn when their payment method is Electronic Check
```

```python
# 5 Finding the number of churn customer with respect to Streaming TV plan
```

```python
churn_by_tv_streaming=df[df['Churn']=='Yes'].groupby('StreamingTV')['customerID'].count()

plt.pie(churn_by_tv_streaming, labels=churn_by_tv_streaming.index, autopct='%.0f%%')
plt.show()
```

```python
# Conclusion - The above pie chart reveals that customer with no TV streaming in their plans tends to churn out more likely
```

```python
plt.pie(churn_by_tenure, labels=churn_by_tenure.index, autopct='%.0f%%')
plt.show()
```

```python
# 6 Finding the churn pattern of the customers with respect to online security in their plan.
```

```python
churn_by_onlinesecurity=df[df['Churn']=='Yes'].groupby('OnlineSecurity')['customerID'].count()

plt.pie(churn_by_onlinesecurity, labels=churn_by_onlinesecurity.index, autopct='%.0f%%')
plt.show()
```

```python
# Conclusion - Customers with no online security in their plan tends to churn out very likely
```

```python
# Finding the churn pattern with respect to Internet Service type
```

```python
churn_by_typeof_internetService=df[df['Churn']=='Yes'].groupby('InternetService')['customerID'].count()

plt.pie(churn_by_typeof_internetService, labels=churn_by_typeof_internetService.index, autopct='%.0f%%')
plt.show()
```
